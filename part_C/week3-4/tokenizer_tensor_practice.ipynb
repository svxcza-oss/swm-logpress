```python
model_name = "Qwen/Qwen2.5-0.5B-Instruct"

tokenizer = AutoTokenizer.from_pretrained(
    model_name
)

text = """INFO server started
WARN memory usage high
ERROR database timeout"""

encoded = tokenizer(text)
print(type(tokenizer))
print(encoded["input_ids"])
print(encoded["attention_mask"])

```

```python
ids = tokenizer.encode(text)
print(ids)
decoded_text = tokenizer.decode(ids)
print(decoded_text)
```

```python
input_Ids =tokenizer.encode(text)
print("input_ids")
print(input_ids)

tokens = tokenizer.convert_ids_to_tokens(input_ids)

print("\n tokens")
print(tokens)

decoded = tokenizer.decode(
    input_ids
)

print("\n decoded:")
print(decoded)

print("\nspecial tokens")

print("BOS token:", tokenizer.bos_token)
print("EOS token:", tokenizer.eos_token)
print("PAD token:", tokenizer.pad_token)
print("UNK token:", tokenizer.unk_token)

print("\nSpecial Token IDs")

print("BOS token id:", tokenizer.bos_token_id)
print("EOS token id:", tokenizer.eos_token_id)
print("PAD token id:", tokenizer.pad_token_id)
print("UNK token id:", tokenizer.unk_token_id)

#BOS
#→ Begin Of Sequence

#EOS
#→ End Of Sequence

#PAD
#→ Padding

#UNK
#→ Unknown
```

```python
text = """INFO server started
WARN memory usage high
ERROR database timeout"""

encoded = tokenizer(
    text,
    return_offsets_mapping = True,
    add_special_tokens = False
)
input_ids = encoded["input_ids"]
offsets = encoded["offset_mapping"]

tokens = tokenizer.convert_ids_to_tokens(input_ids)
for token_id,token ,(start,end) in zip( input_ids,
    tokens,
    offsets):
  original_text = text[start:end]
  print(token_id,start,end,token,
  repr(original_text)
    )
```

```python
#softmax : score값을 비율로 변환

Q = torch.randn(2,4,10,64)
K = torch.randn(2,4,10,64)

scores = Q @ K.transpose(-2,-1)

weights = torch.softmax(
    scores,
    dim= -1
)

print(scores.shape)
print(weights.shape)

print(weights.sum(dim=-1))
```

```python
keys = torch.randn(2, 8, 10, 64)

scores = torch.tensor([
    0.1,
    0.7,
    0.2,
    0.9,
    0.3,
    0.8,
    0.4,
    0.6,
    0.5,
    0.05
])
values,indices = torch.topk(
    scores, k =4
)

print("values :", values)
print("indices:", indices)

selected_keys= torch.index_select(
    keys,
    dim = 2,
    index = indices
)
print("원래 keys:", keys.shape)
print("선택 keys:", selected_keys.shape)
```

```python
#gather : 각 행마다 서로 다른 index를 줄 수있음
scores = torch.tensor([
    [
        [0.1, 0.7, 0.2, 0.9, 0.3],
        [0.8, 0.2, 0.6, 0.1, 0.5]
    ]
])

values, indices = torch.topk(
    scores,
    k=2,
    dim = -1
)
print(values)
print(indices)
print(indices.shape)
```
