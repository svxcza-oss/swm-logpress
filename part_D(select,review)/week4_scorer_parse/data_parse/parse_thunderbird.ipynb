```python
import csv
import sys

INPUT = "/Users/mac/Downloads/Thunderbird_2k.log"
OUTPUT = "Thunderbird_parsed_2k.csv"

GZIP = False
LINE_ID_START = 0
PROGRESS_EVERY = 5_000_000

fields = ["line_id", "label", "timestamp", "date", "user", "month", "day", "time", "location", "component", "content"]

def parse_line(line):
    p = line.rstrip("\n").split(None, 9)
    
    if len(p) < 9:
        d = {k: ""  for k in fields}
        d["content"] = line.rstrip("\n")
        return d
    return {
        "line_id": None,
        "label": p[0],
        "timestamp" : p[1],
        "date" : p[2],
        "user" : p[3],
        "month" : p[4],
        "day" : p[5],
        "time" : p[6],
        "location" : p[7],
        "component" : p[8].rstrip(":"),
        "content" : p[9] if len(p) > 9 else "",

    }
    
def open_out(path, gzip_on):
    if gzip_on:
        import gzip
        return gzip.open(path + ".gz", "wt", newline="", encoding="utf-8")
    return open(path, "w", newline="", encoding="utf-8")

def main():
    fin = open(INPUT, encoding="utf-8-sig", errors="ignore")
    fout = open_out(OUTPUT, GZIP)
    w = csv.DictWriter(fout, fieldnames=fields)
    w.writeheader()
    
    i = -1
    for i,line in enumerate(fin):
        row = parse_line(line)
        row["line_id"] = i + LINE_ID_START
        w.writerow(row)
        if i % PROGRESS_EVERY == 0:
            print(f"{i:,} 줄 처리...", flush=True)
            
    fin.close(); fout.close()
    print(f"완료: 총 {i+1:,} 줄 -> {OUTPUT}{'.gz' if GZIP else ''}")
    

if __name__ == "__main__":
    main()
```
